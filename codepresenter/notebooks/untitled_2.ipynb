```python
import matplotlib.pyplot as pltx = [1, 2, 3, 4, 5]y = [2, 4, 6, 8, 10]plt.plot(x, y)plt.title("Simple Line Plot")plt.xlabel("X values")plt.ylabel("Y values")plt.show()
```

```python
import matplotlib.pyplot as pltcategories = ["A", "B", "C", "D"]values = [5, 7, 3, 8]plt.bar(categories, values)plt.title("Bar Chart")plt.xlabel("Category")plt.ylabel("Value")plt.show()
```
